Create a separate copy of the raw customers dataset. This keeps the original data unchanged while we perform cleaning.

Remove leading and trailing spaces from city names and convert them to title case so that differently formatted names become consistent.

```python
customers_clean["city"] = (
    customers_clean["city"]
    .astype("string")
    .str.strip()
    .str.title()
)

print("Unique cities:", customers_clean["city"].nunique())
print("\nCity counts:")
print(customers_clean["city"].value_counts().sort_index())
```

Standardise gender values into three categories: male, female, and unknown. Blank values and unrecognised spellings will be mapped to unknown.

```python
gender_mapping = {
    "m": "male",
    "male": "male",
    "man": "male",
    "f": "female",
    "female": "female",
    "woman": "female"
}

customers_clean["gender"] = (
    customers_clean["gender"]
    .astype("string")
    .str.strip()
    .str.lower()
    .map(gender_mapping)
    .fillna("unknown")
)

print("Gender counts:")
print(customers_clean["gender"].value_counts())
```

Convert age values to numeric format. Treat ages below 10 or above 100 as missing values. Then fill all missing ages using the median age of the corresponding city.

```python
customers_clean["age"] = pd.to_numeric(
    customers_clean["age"], errors="coerce"
)

# Replace ages outside the valid range with missing values
customers_clean.loc[
    ~customers_clean["age"].between(10, 100), "age"
] = np.nan

# Fill missing ages using the median age of each city
customers_clean["age"] = customers_clean["age"].fillna(
    customers_clean.groupby("city")["age"].transform("median")
)

print("Missing ages:", customers_clean["age"].isna().sum())
print("Minimum age:", customers_clean["age"].min())
print("Maximum age:", customers_clean["age"].max())
```

Convert the signup_date column into pandas datetime format so that dates can be handled correctly during analysis.

```python
customers_clean["signup_date"] = pd.to_datetime(
    customers_clean["signup_date"], errors="coerce"
)

print("Signup date data type:", customers_clean["signup_date"].dtype)
print("Missing signup dates:", customers_clean["signup_date"].isna().sum())
```

Some orders have multiple records with different statuses. Sort records by order_ts and keep the latest record for each order_id, ensuring earlier pending records are replaced by their final status.

```python
orders_clean = orders_clean.sort_values(
    "order_ts"
)

orders_clean = orders_clean.drop_duplicates(
    subset="order_id",
    keep="last"
).copy()

print("Rows after resolving duplicate order IDs:", len(orders_clean))
print("\nOrder status counts:")
print(orders_clean["status"].value_counts())
```

Convert order_amount into numeric format. Any invalid or non-numeric values will become missing (NaN) so we can calculate and fill them in the next step.

```python
orders_clean["order_amount"] = pd.to_numeric(
    orders_clean["order_amount"],
    errors="coerce"
)

print("Missing order amounts:", orders_clean["order_amount"].isna().sum())
print("Data type:", orders_clean["order_amount"].dtype)
```

Calculate missing order amounts using the product price, quantity, and discount percentage. The formula is: price_inr Ã— quantity Ã— (1 âˆ’ discount_pct / 100). Round the calculated amount to two decimal places.

```python
orders_clean = orders_clean.merge(
    products[["product_id", "price_inr"]],
    on="product_id",
    how="left"
)

calculated_amount = (
    orders_clean["price_inr"]
    * orders_clean["quantity"]
    * (1 - orders_clean["discount_pct"] / 100)
).round(2)

orders_clean["order_amount"] = (
    orders_clean["order_amount"].fillna(calculated_amount)
)

print("Missing order amounts:", orders_clean["order_amount"].isna().sum())
print("Total orders:", len(orders_clean))
```

Add the product category to the cleaned orders dataset using product_id to match records with the products dataset.

```python
orders_clean["category"] = orders_clean["product_id"].map(
    products.set_index("product_id")["category"]
)

print("Missing categories:", orders_clean["category"].isna().sum())
print("Categories:", orders_clean["category"].unique())
```

Convert delivery_minutes to numeric format and replace negative values with missing values. For completed orders, fill missing delivery times using the median delivery time of the corresponding product category. Cancelled orders must retain missing delivery times because they were not completed.

```python
# Convert delivery_minutes to numeric
orders_clean["delivery_minutes"] = pd.to_numeric(
    orders_clean["delivery_minutes"],
    errors="coerce"
)

# Replace negative delivery times with missing values
orders_clean.loc[
    orders_clean["delivery_minutes"] < 0,
    "delivery_minutes"
] = np.nan

# Calculate category-wise median using completed orders only
category_median = (
    orders_clean[orders_clean["status"] == "completed"]
    .groupby("category")["delivery_minutes"]
    .transform("median")
)

# Fill missing delivery times for completed orders
completed_mask = orders_clean["status"] == "completed"

orders_clean.loc[completed_mask, "delivery_minutes"] = (
    orders_clean.loc[completed_mask, "delivery_minutes"]
    .fillna(category_median)
)

print("Missing delivery minutes among completed:",
      orders_clean.loc[completed_mask, "delivery_minutes"].isna().sum())

print("Missing delivery minutes among cancelled:",
      orders_clean.loc[~completed_mask, "delivery_minutes"].isna().sum())
```

Q1 - Identify customers with completed orders where order_amount is greater than INR 500. Group the qualifying orders by customer and select customers whose total qualifying order amount exceeds INR 8,000.

```python
query_1 = """
SELECT
    customer_id,
    COUNT(*) AS qualifying_orders,
    ROUND(SUM(order_amount), 2) AS total_spend
FROM orders
WHERE status = 'completed'
  AND order_amount > 500
GROUP BY customer_id
HAVING SUM(order_amount) > 8000
ORDER BY total_spend DESC;
"""

q1_result = pd.read_sql_query(query_1, conn)

print(q1_result.to_string(index=False))
print("\nTotal qualifying customers:", len(q1_result))
```

Q5 - Use two CTEs:
1. Calculate the total completed order spend for each customer.
2. Calculate the average of those customer totals.
Return customers whose total completed spend is above that average.

```python
query_5 = """
WITH customer_spend AS (
    SELECT
        customer_id,
        SUM(order_amount) AS total_spend
    FROM orders
    WHERE status = 'completed'
    GROUP BY customer_id
),
average_spend AS (
    SELECT AVG(total_spend) AS avg_spend
    FROM customer_spend
)
SELECT
    cs.customer_id,
    ROUND(cs.total_spend, 2) AS total_spend
FROM customer_spend cs
CROSS JOIN average_spend av
WHERE cs.total_spend > av.avg_spend
ORDER BY cs.total_spend DESC;
"""

q5_result = pd.read_sql_query(query_5, conn)

print(q5_result.to_string(index=False))
print("\nTotal customers:", len(q5_result))
```

Q6 - Calculate each customer's total completed order spend. Then use three window functions â€” RANK(), DENSE_RANK(), and ROW_NUMBER() â€” to rank customers within each city by total spend in descending order.

```python
query_6 = """
WITH customer_spend AS (
    SELECT
        c.customer_id,
        c.city,
        SUM(o.order_amount) AS total_spend
    FROM customers c
    JOIN orders o
        ON c.customer_id = o.customer_id
    WHERE o.status = 'completed'
    GROUP BY c.customer_id, c.city
),
ranked_customers AS (
    SELECT
        customer_id,
        city,
        ROUND(total_spend, 2) AS total_spend,

        RANK() OVER (
            PARTITION BY city
            ORDER BY total_spend DESC
        ) AS rank_position,

        DENSE_RANK() OVER (
            PARTITION BY city
            ORDER BY total_spend DESC
        ) AS dense_rank_position,

        ROW_NUMBER() OVER (
            PARTITION BY city
            ORDER BY total_spend DESC
        ) AS row_number_position

    FROM customer_spend
)
SELECT *
FROM ranked_customers
ORDER BY city, rank_position, customer_id;
"""

q6_result = pd.read_sql_query(query_6, conn)

print(q6_result.to_string(index=False))
```

```python
expected_top = {
    "Bengaluru": ("C0082", 12856.20),
    "Chennai": ("C0930", 10151.40),
    "Delhi": ("C0291", 13742.60),
    "Hyderabad": ("C0743", 12730.05),
    "Mumbai": ("C0693", 14766.55),
    "Pune": ("C0426", 10377.10)
}

top_customers = (
    q6_result.sort_values("rank_position")
    .groupby("city")
    .first()
)

for city, (customer_id, spend) in expected_top.items():
    actual = top_customers.loc[city]
    assert actual["customer_id"] == customer_id
    assert round(actual["total_spend"], 2) == spend

different_ranks = (
    q6_result["rank_position"] != q6_result["row_number_position"]
).sum()

print("Top customer per city: CHECK PASSED")
print("Rows where RANK differs from ROW_NUMBER:", different_ranks)

assert different_ranks == 8

print("Q6 CHECK PASSED")
```

Q7 - Calculate daily revenue from completed orders. Extract the date from order_ts using its first ten characters, then calculate cumulative revenue using SUM() OVER (ORDER BY day).

```python
query_7 = """
WITH daily_revenue AS (
    SELECT
        SUBSTR(order_ts, 1, 10) AS day,
        SUM(order_amount) AS revenue
    FROM orders
    WHERE status = 'completed'
    GROUP BY SUBSTR(order_ts, 1, 10)
)
SELECT
    day,
    ROUND(revenue, 2) AS daily_revenue,
    ROUND(
        SUM(revenue) OVER (ORDER BY day),
        2
    ) AS cumulative_revenue
FROM daily_revenue
ORDER BY day;
"""

q7_result = pd.read_sql_query(query_7, conn)

print(q7_result.to_string(index=False))
```

Analyse the distribution of completed order amounts using mean, median, quartiles, skewness and the Interquartile Range (IQR). Identify upper and lower outliers using the 1.5 Ã— IQR rule.

```python
amount = completed_orders["order_amount"]

q1 = amount.quantile(0.25)
q3 = amount.quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

distribution = {
    "Mean": round(amount.mean(), 2),
    "Median": round(amount.median(), 2),
    "Skewness": round(amount.skew(), 2),
    "Q1": round(q1, 2),
    "Q3": round(q3, 2),
    "Lower fence": round(lower_fence, 2),
    "Upper fence": round(upper_fence, 2),
    "Upper outliers": int((amount > upper_fence).sum()),
    "Lower outliers": int((amount < lower_fence).sum())
}

for key, value in distribution.items():
    print(f"{key}: {value}")
```
